```python
# ------------------------------------------------------------------------------
# 1. CORE PYTHON & DATA TYPES
# ------------------------------------------------------------------------------
```

```python
# Primitive Data Types
student_id = 101                 # int
test_score = 84.5                # float
student_name = "Pradeep Singh"         # str
is_certified = True              # bool
```

```python
# Non-Primitive Data Structures
raw_scores = [72, 85, 90, 68, 85, 95, 78]        # List (mutable, ordered)
fixed_weights = (0.2, 0.3, 0.5)                   # Tuple (immutable, ordered)
unique_departments = {"Analytics", "Tech", "HR"}  # Set (unique elements)
student_profile = {                               # Dictionary (key-value pairs)
    "id": student_id,
    "name": student_name,
    "active": is_certified,
    "scores": raw_scores
}

print("=== Profile Created Successfully ===")
print(f"Name: {student_profile['name']} | ID: {student_profile['id']}")
print(f"Raw Scores: {student_profile['scores']}")
```

```python
# Function with Type Annotations and Docstring (Industry Best Practice)
def calculate_grade(score: float) -> str:
    """Classifies a numeric score into standardized academic grades."""
    if score >= 90:
        return "Grade A"
    elif score >= 75:
        return "Grade B"
    elif score >= 60:
        return "Grade C"
    else:
        return "Needs Improvement"

print(f"Sample Grade for {test_score}: {calculate_grade(test_score)}")
```

```python
# ------------------------------------------------------------------------------
# 2. NUMPY & VECTORIZED COMPUTATION
# ------------------------------------------------------------------------------

import numpy as np

scores_arr = np.array(raw_scores)
```

```python
# Vectorized operation example (Scaling scores by 5% curve)
curved_scores = scores_arr * 1.05
print("\n=== Vectorized Scaling via NumPy ===")
print("Original Scores:", scores_arr)
print("Curved Scores (rounded):", np.round(curved_scores, 2))
```

```python
# ------------------------------------------------------------------------------
# 3. STATISTICAL FOUNDATIONS IMPLEMENTATION
# ------------------------------------------------------------------------------
# Calculating central tendency and dispersion from scratch and with libraries
mean_val = np.mean(scores_arr)
median_val = np.median(scores_arr)
std_dev = np.std(scores_arr)
variance_val = np.var(scores_arr)
```

```python
# Interquartile Range (IQR) for Outlier Sensitivity
q75, q25 = np.percentile(scores_arr, [75, 25])
iqr_val = q75 - q25

print("\n=== Statistical Summary ===")
print(f"Mean (Central Tendency)    : {mean_val:.2f}")
print(f"Median (Robust Center)     : {median_val:.2f}")
print(f"Variance                   : {variance_val:.2f}")
print(f"Standard Deviation (Spread): {std_dev:.2f}")
print(f"IQR (Spread of Middle 50%) : {iqr_val:.2f}")
```

```python
# ------------------------------------------------------------------------------
# 4. DATA INTEGRITY & BASIC DATAFRAME HANDLING
# ------------------------------------------------------------------------------
import pandas as pd

dataset = {
    "Candidate_ID": [201, 202, 203, 204, 205, 206],
    "Age": [22, 24, np.nan, 29, 23, 35],
    "Assessment_Score": [78, 88, 92, np.nan, 65, 84],
    "Category": ["General", "Technical", "Technical", "Operations", "General", "Technical"]
}

df = pd.DataFrame(dataset)
```

```python
# Imputation (Median for Age, Mean for Assessment_Score)
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Assessment_Score"] = df["Assessment_Score"].fillna(df["Assessment_Score"].mean())

print("\n=== Cleaned Candidate Dataset ===")
print(df)
```
