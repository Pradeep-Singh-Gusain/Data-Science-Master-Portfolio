```python
import getpass
import pandas as pd
from sqlalchemy import create_engine, text

# 1. Secure Authentication
db_password = getpass.getpass("Enter your PostgreSQL password: ")
engine = create_engine(f'postgresql://postgres:{db_password}@localhost:2709/zomato_db')

# 2. Strict Schema Definition (Data Integrity Best Practice)
create_table_query = """
DROP TABLE IF EXISTS zomato_restaurants;

CREATE TABLE zomato_restaurants (
    url TEXT,
    address TEXT,
    name TEXT,
    online_order VARCHAR(10),
    book_table VARCHAR(10),
    rate VARCHAR(20),
    votes INT,
    phone TEXT,
    location TEXT,
    rest_type TEXT,
    dish_liked TEXT,
    cuisines TEXT,
    approx_cost_for_two VARCHAR(50),
    reviews_list TEXT,
    menu_item TEXT,
    listed_in_type TEXT,
    listed_in_city TEXT
);
"""

with engine.begin() as conn:
    conn.execute(text(create_table_query))
print(">> Clean enterprise table schema successfully initialized.")

# 3. Batch Streaming with Explicit Mapping
file_path = '../../03_Real_World_Dataset_Project/zomato.csv'
chunksize = 50000
batch_count = 1

for chunk in pd.read_csv(file_path, chunksize=chunksize, low_memory=False):
    # Explicit mapping to match strict table schema
    chunk = chunk.rename(columns={
        'approx_cost(for two people)': 'approx_cost_for_two',
        'listed_in(type)': 'listed_in_type',
        'listed_in(city)': 'listed_in_city'
    })
    
    # Append chunk to pre-defined table
    chunk.to_sql('zomato_restaurants', engine, if_exists='append', index=False)
    print(f">> Batch {batch_count} successfully streamed and inserted.")
    batch_count += 1

print("\n[SUCCESS]: Zomato dataset enterprise ingestion pipeline completed without anomalies!")
```
